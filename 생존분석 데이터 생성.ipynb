```python
import random
import pandas as pd
import numpy as np
from faker import Faker
from datetime import datetime, timedelta
```

```python
fake = Faker()
```

```python

# 생성할 데이터 개수
num_records = 20000

# 치료 유형 목록
treatment_types = ["Chemotherapy", "Radiation", "Surgery", "Medication"]

# 동반 질환 목록
comorbidity_options = ["Hypertension", "Diabetes", "Heart Disease", "Asthma", "None"]

# 데이터 저장할 리스트
data = []
```

```python
for _ in range(num_records):
    patient_id = _ + 1
    age = random.randint(30, 90)  # 30~90세 사이의 환자
    gender = random.choice(["Male", "Female"])
    
    # 진단 날짜 (최근 10년 내)
    diagnosis_date = fake.date_between(start_date="-10y", end_date="today")
    
    # 치료 유형
    treatment_type = random.choice(treatment_types)
    
    # 동반 질환
    comorbidities = random.choice(comorbidity_options)
    
    # 생존 분석을 위한 이벤트(사망/재발 여부) 및 생존 시간 설정
    event = random.choice([0, 1])  # 0 = 생존, 1 = 이벤트 발생(사망, 재발 등)

     #수치형 변수
    bmi = round(random.uniform(18.5, 35.0), 1)  # BMI (18.5~35.0)
    blood_pressure = f"{random.randint(90, 180)}/{random.randint(60, 110)}"  # 혈압 (수축기/이완기)
    cholesterol = random.randint(130, 300)  # 콜레스테롤 수치 (130~300 mg/dL)
    medication_dose = round(random.uniform(5, 100), 2) if treatment_type == "Medication" else 0  # 약물 투여량 (5~100mg)

    # 코호트 컬럼 (이진 변수)
    smoking_status = np.random.choice([0, 1])  # 0: 비흡연자, 1: 흡연자
    exercise_level = np.random.choice([0, 1])  # 0: 거의 안 함, 1: 주 3회 이상
    family_history_cvd = np.random.choice([0, 1])  # 0: 가족력 없음, 1: 가족력 있음

    # 생존 시간 설정 (진단일 이후 몇 개월까지 생존했는지)
    if event == 1:
        survival_time = random.randint(1, 120)  # 1~120개월 (10년 이내)
    else:
        survival_time = random.randint(121, 180)  # 10~15년 이상 생존
    
    data.append([
        patient_id, age, gender, diagnosis_date, treatment_type, comorbidities, event, survival_time,
        bmi, blood_pressure, cholesterol, medication_dose, smoking_status, exercise_level, family_history_cvd
    ])
```

```python
# DataFrame 생성
columns = [
    "Patient_ID", "Age", "Gender", "Diagnosis_Date", "Treatment_Type", "Comorbidities", "Event", "Survival_Time",
    "BMI", "Blood_Pressure", "Cholesterol", "Medication_Dose", "Smoking_Status", "Exercise_Level", "Family_History_CVD"
]

survival_df = pd.DataFrame(data, columns=columns)
```

```python

# CSV 저장
today = datetime.today().date()
save_path = 'C:/Users/WoojaeYang/Downloads/'
survival_df.to_csv(f"{save_path}survival_analysis_data_{today}.csv", index=False)
```
