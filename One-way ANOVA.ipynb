```python
seer2 = main4[main4['SEER_GRP'] == 2]
seer9 = main4[main4['SEER_GRP'] == 9]
seer7 = main4[main4['SEER_GRP'] == 7]
seer1 = main4[main4['SEER_GRP'] == 1]
```

```python
grp2 = seer2['TIME']
grp9 = seer9['TIME']
grp7 = seer7['TIME']
grp1 = seer1['TIME']
```

```python
f_stat, p_val = stats.f_oneway(grp2, grp9, grp7, grp1)
print(f'F_val= {f_stat:.4f}, P-val= {p_val:.4f}' )
```
