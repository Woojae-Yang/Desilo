```python
import random as rd

fake = Faker()
```

```python
num_records = 10000  # 생성할 데이터 개수
market_segments = ["AUTOMOBILE", "FURNITURE", "HOUSEHOLD", "ELECTRONICS", "CLOTHING"]
```

```python
data = []
```

```python
for _ in range(num_records):
    c_custkey = _
    c_name = fake.name()
    c_address = fake.address()
    c_phone = fake.phone_number()
    c_acctbal = round(rd.uniform(1000, 10000), 2)  # 1000~10000 사이 랜덤 잔고
    c_mktsegment = rd.choice(market_segments)
    c_comment = fake.text(max_nb_chars=100)  # 최대 100자
    
    data.append([c_custkey, c_name, c_address, c_phone, c_acctbal, c_mktsegment, c_comment])
```

```python
from datetime import datetime
today = datetime.today().date()
print(today)
```

```python
# CSV로 저장
save_path = 'C:/Users/WoojaeYang/Downloads/'
with open(f'{save_path}customer_data_{today}.csv', mode='w', newline='', encoding='utf-8') as file:
    writer = csv.writer(file)
    writer.writerow(["C_CUSTKEY", "C_NAME", "C_ADDRESS", "C_PHONE", "C_ACCTBAL", "C_MKTSEGMENT", "C_COMMENT"])
    writer.writerows(data)
```
